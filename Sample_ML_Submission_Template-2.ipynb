# **Project Summary -**

Expense Anomaly Auditor is an unsupervised machine learning project that uses K-Means clustering to automatically flag unusual or suspicious transactions in a financial dataset. Organisations and individuals generate thousands of transactions every month. Manual auditing is slow and inconsistent, and it tends to catch only the most obvious irregularities. This project builds a data-driven auditing assistant that learns what "normal" spending looks like and highlights what deviates from it, so a reviewer can focus on a short, prioritised list instead of the full ledger.

The project starts with a transaction dataset from Kaggle containing fields such as date, amount, category, merchant, and payment type. The raw data is cleaned by handling missing values, removing duplicates, correcting data types, and treating obvious entry errors. Exploratory analysis follows, looking at spending distributions, category-wise patterns, and day-of-week and time-of-day behaviour. Because financial amounts are usually heavily right-skewed, a log transformation is considered at this stage.

Feature engineering is the core of the project. Beyond the raw amount, features such as the amount's z-score within its category, transaction frequency per merchant, time since the previous transaction, weekend or night-time flags, and rolling averages are created. Categorical variables are encoded, and all numeric features are standardised, since K-Means is distance-based and highly sensitive to scale.

K-Means is then applied. The number of clusters (k) is chosen using the elbow method and silhouette score. Anomalies are detected in two complementary ways:

Distance from centroid: transactions that lie unusually far from their assigned cluster centre, for example beyond the 99th percentile of distances within that cluster.
Small clusters: clusters containing very few points, which represent rare spending behaviour.

These signals are combined into a single anomaly score, and transactions are ranked by it.

Since the approach is unsupervised, validation needs care. Flagged transactions are manually inspected and sanity-checked against business logic. If the dataset includes fraud or anomaly labels, they are used only for evaluation (precision, recall) and never for training. Isolation Forest can be added as a baseline to see how K-Means compares.

The final deliverables are visualisations (a 2D PCA cluster plot, an anomaly score distribution, and flagged transactions over time) and an exportable audit table. Optionally, everything is wrapped in a Streamlit app where a user uploads a CSV, adjusts detection sensitivity, and downloads a report.

Through this project you practise data cleaning, feature engineering, clustering, unsupervised model evaluation, and communicating findings to a non-technical audience.

## Library Rationale

### 1. Core Data Handling

| Library | Role in this project |
|---|---|
| **pandas** | Stores transactions in a DataFrame. Used for loading CSVs, cleaning, date handling, `groupby` aggregations (such as average spend per category), and building every engineered feature. |
| **numpy** | Fast numerical operations: log transforms for skewed amounts, distance calculations between points and centroids, and percentile thresholds for flagging. |
| **scipy.stats** | Statistical utilities such as `zscore`, used to measure how far a transaction's amount sits from its category's typical spend. |

### 2. Visualisation

| Library | Role in this project |
|---|---|
| **matplotlib** | The foundation for plotting. Used for the elbow curve, silhouette plot, and histograms. |
| **seaborn** | Statistical plots with minimal code: boxplots by category, distribution plots, and correlation heatmaps. |
| **plotly.express** | Interactive charts with hover details, ideal for the PCA cluster plot and the Streamlit app, where a reviewer wants to inspect individual transactions. |

### 3. Data Acquisition

| Library | Role in this project |
|---|---|
| **kagglehub** | Downloads Kaggle datasets directly from the notebook, which keeps the data step reproducible without manual downloads. |

### 4. Preprocessing

| Library | Why it matters |
|---|---|
| **StandardScaler** | K-Means is distance-based, so a feature measured in thousands (amount) would drown out a binary feature (weekend flag). Scaling gives every feature a mean of 0 and a standard deviation of 1 so each contributes fairly. |
| **OneHotEncoder** | Converts categories such as "Food" or "Travel" into separate 0/1 columns. This avoids implying a false order (Food = 1, Travel = 2) that would distort distances. `pd.get_dummies` is an equivalent alternative. |

### 5. Clustering and Anomaly Detection

| Library | Why it matters |
|---|---|
| **KMeans** | The core model. It groups transactions into *k* clusters of similar behaviour. Transactions far from their centroid, or sitting in very small clusters, become anomaly candidates. |
| **IsolationForest** | A comparison baseline that detects anomalies using different logic: how quickly random splits can isolate a point. Agreement or disagreement with K-Means shows how trustworthy your flags are. |

### 6. Dimensionality Reduction

| Library | Why it matters |
|---|---|
| **PCA** | Your feature matrix has many dimensions and cannot be drawn directly. PCA compresses it to two components for a scatter plot of clusters and anomalies. It is for visualisation only: cluster on the full scaled features, not the PCA output. |

### 7. Evaluation

| Library | Why it matters |
|---|---|
| **silhouette_score** | Measures how well separated the clusters are, on a scale from -1 to 1. Used alongside the elbow method to choose *k*. |
| **precision_score / recall_score** | Used only when labels exist, or when you inject synthetic anomalies. Precision shows how many flags were genuine; recall shows how many genuine anomalies you caught. |

### 8. Model Persistence

| Library | Why it matters |
|---|---|
| **joblib** | Saves the fitted scaler and K-Means model to disk so the Streamlit app can load them instead of retraining on every run. |

### 9. Application Layer

| Library | Why it matters |
|---|---|
| **streamlit** | Turns the analysis into a web app with a file uploader, sensitivity slider, results table, and download button, with no frontend code required. |

### 10. Housekeeping

| Library | Why it matters |
|---|---|
| **warnings** | Suppresses non-critical warnings (such as pandas deprecation notices) to keep notebook output readable. Remove this line while debugging so you don't miss useful messages. |
