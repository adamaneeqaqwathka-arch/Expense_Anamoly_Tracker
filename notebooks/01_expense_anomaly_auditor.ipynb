# **Project Summary -**

Expense Anomaly Auditor is an unsupervised machine learning project that uses K-Means clustering to automatically flag unusual or suspicious transactions in a financial dataset. Organisations and individuals generate thousands of transactions every month. Manual auditing is slow and inconsistent, and it tends to catch only the most obvious irregularities. This project builds a data-driven auditing assistant that learns what "normal" spending looks like and highlights what deviates from it, so a reviewer can focus on a short, prioritised list instead of the full ledger.

The project starts with a transaction dataset from Kaggle containing fields such as date, amount, category, merchant, and payment type. The raw data is cleaned by handling missing values, removing duplicates, correcting data types, and treating obvious entry errors. Exploratory analysis follows, looking at spending distributions, category-wise patterns, and day-of-week and time-of-day behaviour. Because financial amounts are usually heavily right-skewed, a log transformation is considered at this stage.

Feature engineering is the core of the project. Beyond the raw amount, features such as the amount's z-score within its category, transaction frequency per merchant, time since the previous transaction, weekend or night-time flags, and rolling averages are created. Categorical variables are encoded, and all numeric features are standardised, since K-Means is distance-based and highly sensitive to scale.

K-Means is then applied. The number of clusters (k) is chosen using the elbow method and silhouette score. Anomalies are detected in two complementary ways:

Distance from centroid: transactions that lie unusually far from their assigned cluster centre, for example beyond the 99th percentile of distances within that cluster.
Small clusters: clusters containing very few points, which represent rare spending behaviour.

These signals are combined into a single anomaly score, and transactions are ranked by it.

Since the approach is unsupervised, validation needs care. Flagged transactions are manually inspected and sanity-checked against business logic. If the dataset includes fraud or anomaly labels, they are used only for evaluation (precision, recall) and never for training. Isolation Forest can be added as a baseline to see how K-Means compares.

The final deliverables are visualisations (a 2D PCA cluster plot, an anomaly score distribution, and flagged transactions over time) and an exportable audit table. Optionally, everything is wrapped in a Streamlit app where a user uploads a CSV, adjusts detection sensitivity, and downloads a report.

Through this project you practise data cleaning, feature engineering, clustering, unsupervised model evaluation, and communicating findings to a non-technical audience.

## Library Rationale

### 1. Core Data Handling

| Library | Role in this project |
|---|---|
| **pandas** | Stores transactions in a DataFrame. Used for loading CSVs, cleaning, date handling, `groupby` aggregations (such as average spend per category), and building every engineered feature. |
| **numpy** | Fast numerical operations: log transforms for skewed amounts, distance calculations between points and centroids, and percentile thresholds for flagging. |
| **scipy.stats** | Statistical utilities such as `zscore`, used to measure how far a transaction's amount sits from its category's typical spend. |

### 2. Visualisation

| Library | Role in this project |
|---|---|
| **matplotlib** | The foundation for plotting. Used for the elbow curve, silhouette plot, and histograms. |
| **seaborn** | Statistical plots with minimal code: boxplots by category, distribution plots, and correlation heatmaps. |
| **plotly.express** | Interactive charts with hover details, ideal for the PCA cluster plot and the Streamlit app, where a reviewer wants to inspect individual transactions. |

### 3. Data Acquisition

| Library | Role in this project |
|---|---|
| **kagglehub** | Downloads Kaggle datasets directly from the notebook, which keeps the data step reproducible without manual downloads. |

### 4. Preprocessing

| Library | Why it matters |
|---|---|
| **StandardScaler** | K-Means is distance-based, so a feature measured in thousands (amount) would drown out a binary feature (weekend flag). Scaling gives every feature a mean of 0 and a standard deviation of 1 so each contributes fairly. |
| **OneHotEncoder** | Converts categories such as "Food" or "Travel" into separate 0/1 columns. This avoids implying a false order (Food = 1, Travel = 2) that would distort distances. `pd.get_dummies` is an equivalent alternative. |

### 5. Clustering and Anomaly Detection

| Library | Why it matters |
|---|---|
| **KMeans** | The core model. It groups transactions into *k* clusters of similar behaviour. Transactions far from their centroid, or sitting in very small clusters, become anomaly candidates. |
| **IsolationForest** | A comparison baseline that detects anomalies using different logic: how quickly random splits can isolate a point. Agreement or disagreement with K-Means shows how trustworthy your flags are. |

### 6. Dimensionality Reduction

| Library | Why it matters |
|---|---|
| **PCA** | Your feature matrix has many dimensions and cannot be drawn directly. PCA compresses it to two components for a scatter plot of clusters and anomalies. It is for visualisation only: cluster on the full scaled features, not the PCA output. |

### 7. Evaluation

| Library | Why it matters |
|---|---|
| **silhouette_score** | Measures how well separated the clusters are, on a scale from -1 to 1. Used alongside the elbow method to choose *k*. |
| **precision_score / recall_score** | Used only when labels exist, or when you inject synthetic anomalies. Precision shows how many flags were genuine; recall shows how many genuine anomalies you caught. |

### 8. Model Persistence

| Library | Why it matters |
|---|---|
| **joblib** | Saves the fitted scaler and K-Means model to disk so the Streamlit app can load them instead of retraining on every run. |

### 9. Application Layer

| Library | Why it matters |
|---|---|
| **streamlit** | Turns the analysis into a web app with a file uploader, sensitivity slider, results table, and download button, with no frontend code required. |

### 10. Housekeeping

| Library | Why it matters |
|---|---|
| **warnings** | Suppresses non-critical warnings (such as pandas deprecation notices) to keep notebook output readable. Remove this line while debugging so you don't miss useful messages. |

### Dataset Loading

````python
# Download the dataset from Kaggle (it is cached after the first run).
# First use needs a Kaggle token: kaggle.com -> Settings -> API -> Create New Token.
import glob
import os

DATASET = "priyamchoksi/credit-card-transactions-dataset"

try:
    path = kagglehub.dataset_download(DATASET)
except Exception as err:
    raise RuntimeError(
        f"Could not download {DATASET}. Check your internet connection and Kaggle token. Original error: {err}"
    )
print("Path to dataset files:", path)

# Look for the CSV inside the download folder instead of hard-coding a file name
csv_files = sorted(glob.glob(os.path.join(path, "**", "*.csv"), recursive=True),
                   key=os.path.getsize, reverse=True)
if not csv_files:
    raise FileNotFoundError(f"No CSV file found under {path}")
for f in csv_files:
    print(f"{os.path.basename(f)}  ({os.path.getsize(f) / 1e6:.1f} MB)")
csv_path = csv_files[0]   # the largest CSV is used
````

````python
# Load the data. The full file has about 1.85 million rows, so while developing we read a
# random sample. Set SAMPLE_ROWS = None to load everything (needs several GB of RAM).
SAMPLE_ROWS = 200_000

with open(csv_path, "rb") as fh:
    total_rows = sum(1 for _ in fh) - 1   # lines in the file minus the header line

if SAMPLE_ROWS is None or SAMPLE_ROWS >= total_rows:
    df = pd.read_csv(csv_path)
else:
    rng = np.random.default_rng(42)       # fixed seed so the same sample is drawn every run
    keep_fraction = SAMPLE_ROWS / total_rows
    # skiprows runs once per line; line 0 is the header and is always kept
    df = pd.read_csv(csv_path, skiprows=lambda i: i > 0 and rng.random() > keep_fraction)

print(f"Rows in the full file: {total_rows:,}")
print(f"Rows loaded for analysis: {len(df):,}")
````

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# ============================================================
# Section 3: Data Wrangling
# Every step prints what it found and what it changed.
# ============================================================
import os

df_clean = df.copy()                      # the raw df stays untouched
steps = []                                # collects (step, rows_removed, note) for the summary table

def log(step, removed, note):
    steps.append({"step": step, "rows_removed": removed, "note": note})

# ---- 3.0 Column roles ------------------------------------------------------
# The first name found in your data is used. Edit the lists if your columns are called something else.
def pick(*names):
    return next((n for n in names if n in df_clean.columns), None)

COL_DATE     = pick("trans_date_trans_time", "TransactionDate", "Transaction_Date", "Date", "date", "timestamp")
COL_AMOUNT   = pick("amt", "amount", "Amount", "TransactionAmount", "Transaction_Amount")
COL_CATEGORY = pick("category", "Category", "MerchantCategory", "Merchant_Category")
COL_MERCHANT = pick("merchant", "Merchant", "merchant_name", "MerchantName")
COL_TXN_ID   = pick("trans_num", "transaction_id", "TransactionID", "Transaction_ID")
COL_LABEL    = pick("is_fraud", "isFraud", "Class", "fraud", "Fraud")   # evaluation only, never a feature

roles = dict(date=COL_DATE, amount=COL_AMOUNT, category=COL_CATEGORY,
             merchant=COL_MERCHANT, txn_id=COL_TXN_ID, label=COL_LABEL)
print("Column roles detected:")
for role, col in roles.items():
    print(f"  {role:<9} -> {col}")
assert COL_AMOUNT is not None, "No amount column found. Set COL_AMOUNT by hand."

# ---- 3.1 Drop leftover index columns ---------------------------------------
index_cols = [c for c in df_clean.columns if c.startswith("Unnamed:")]
if index_cols:
    df_clean = df_clean.drop(columns=index_cols)
    print(f"\nDropped index column(s): {index_cols}")

# ---- 3.2 Fix data types ----------------------------------------------------
print("\n--- Data types ---")
if COL_DATE:
    before_na = df_clean[COL_DATE].isna().sum()
    df_clean[COL_DATE] = pd.to_datetime(df_clean[COL_DATE], errors="coerce")
    new_na = df_clean[COL_DATE].isna().sum() - before_na
    print(f"{COL_DATE}: converted to datetime ({new_na:,} values could not be parsed)")

if not pd.api.types.is_numeric_dtype(df_clean[COL_AMOUNT]):
    cleaned = df_clean[COL_AMOUNT].astype(str).str.replace(r"[^\d.\-]", "", regex=True)
    df_clean[COL_AMOUNT] = pd.to_numeric(cleaned, errors="coerce")
    print(f"{COL_AMOUNT}: text converted to numbers")
else:
    print(f"{COL_AMOUNT}: already numeric")

text_cols = df_clean.select_dtypes(include=["object", "string"]).columns
for c in text_cols:
    df_clean[c] = df_clean[c].astype("string").str.strip()
print(f"Stripped extra spaces in {len(text_cols)} text columns")

# Some versions of this dataset prefix every merchant with "fraud_". It carries no meaning.
if COL_MERCHANT and df_clean[COL_MERCHANT].str.startswith("fraud_", na=False).mean() > 0.9:
    df_clean[COL_MERCHANT] = df_clean[COL_MERCHANT].str.replace(r"^fraud_", "", regex=True)
    print(f"{COL_MERCHANT}: removed the 'fraud_' prefix")

# ---- 3.3 Missing values ----------------------------------------------------
print("\n--- Missing values ---")
# Rows with no amount or no date cannot be scored, so they are dropped.
must_have = [c for c in (COL_AMOUNT, COL_DATE) if c]
n0 = len(df_clean)
df_clean = df_clean.dropna(subset=must_have)
log("Missing amount or date", n0 - len(df_clean), f"dropped rows with no {must_have}")

# Other columns: fill instead of dropping, so good rows are not thrown away.
for c in df_clean.columns:
    n_missing = df_clean[c].isna().sum()
    if n_missing == 0 or c == COL_LABEL:
        continue
    if pd.api.types.is_numeric_dtype(df_clean[c]):
        df_clean[c] = df_clean[c].fillna(df_clean[c].median())
        print(f"{c}: {n_missing:,} missing -> filled with the median")
    elif pd.api.types.is_datetime64_any_dtype(df_clean[c]):
        pass
    else:
        df_clean[c] = df_clean[c].fillna("Unknown")
        print(f"{c}: {n_missing:,} missing -> filled with 'Unknown'")
if len(steps) and steps[-1]["rows_removed"] == 0 and df_clean.isna().sum().sum() == 0:
    print("No missing values remain.")

# ---- 3.4 Duplicates --------------------------------------------------------
print("\n--- Duplicates ---")
n0 = len(df_clean)
if COL_TXN_ID:
    df_clean = df_clean.drop_duplicates(subset=[COL_TXN_ID])
    note = f"same {COL_TXN_ID} appearing twice"
else:
    df_clean = df_clean.drop_duplicates()
    note = "rows identical in every column"
log("Duplicates", n0 - len(df_clean), note)
print(f"Removed {n0 - len(df_clean):,} duplicates ({note})")

# ---- 3.5 Impossible values -------------------------------------------------
print("\n--- Impossible values ---")
n0 = len(df_clean)
non_positive = (df_clean[COL_AMOUNT] <= 0).sum()
df_clean = df_clean[df_clean[COL_AMOUNT] > 0]
log("Amount <= 0", n0 - len(df_clean), "refunds and zero amounts cannot be log-transformed")
print(f"Amounts <= 0 removed: {non_positive:,}")

if COL_DATE:
    n0 = len(df_clean)
    df_clean = df_clean[df_clean[COL_DATE] <= pd.Timestamp.now()]
    log("Future dates", n0 - len(df_clean), "date is after today")
    print(f"Future-dated rows removed: {n0 - len(df_clean):,}")

# ---- 3.6 Outliers: look, but do not delete ---------------------------------
print("\n--- Outliers (kept on purpose) ---")
q = df_clean[COL_AMOUNT].quantile([0.5, 0.99, 0.999])
print(f"Median {q[0.5]:,.2f} | 99th percentile {q[0.99]:,.2f} | 99.9th percentile {q[0.999]:,.2f} | max {df_clean[COL_AMOUNT].max():,.2f}")
print("Large amounts are what the detector should find, so they stay in the data.")

# ---- 3.7 Summary and save --------------------------------------------------
df_clean = df_clean.reset_index(drop=True)
print("\n=== Wrangling summary ===")
print(f"Rows before: {len(df):,}   Rows after: {len(df_clean):,}   Removed: {len(df) - len(df_clean):,} "
      f"({(len(df) - len(df_clean)) / len(df):.2%})")
display(pd.DataFrame(steps))
if COL_LABEL:
    print(f"\nLabel column '{COL_LABEL}' kept for evaluation only. Do not use it as a feature.")

os.makedirs("data", exist_ok=True)
df_clean.to_csv("data/clean.csv", index=False)
print("Saved: data/clean.csv")
df_clean.info()
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**H0 (null):** The mean log-amount of weekend transactions equals that of weekday transactions.

**H1 (alternate):** The mean log-amount of weekend transactions differs from that of weekday transactions.

#### 2. Perform an appropriate statistical test.

````python
ALPHA = 0.05   # significance level: below this p-value we reject the null hypothesis

# H1: do weekend transactions have a different average amount than weekday ones?
# Compare on log(1 + amount): the raw amounts are too skewed for a mean comparison.
if "is_weekend" in viz:
    weekday = viz.loc[viz["is_weekend"] == 0, "log_amt"]
    weekend = viz.loc[viz["is_weekend"] == 1, "log_amt"]
    t_stat, p_value = stats.ttest_ind(weekday, weekend, equal_var=False)       # Welch t-test: no equal-variance assumption
    pooled_sd = np.sqrt((weekday.var() + weekend.var()) / 2)
    cohens_d = (weekend.mean() - weekday.mean()) / pooled_sd                   # effect size: how big is the gap?
    print(f"Weekday: n={len(weekday):,}, mean log-amount={weekday.mean():.4f}")
    print(f"Weekend: n={len(weekend):,}, mean log-amount={weekend.mean():.4f}")
    print(f"t = {t_stat:.3f}, p-value = {p_value:.4g}, Cohen's d = {cohens_d:.3f}")
    print("Conclusion:", "reject H0 (the means differ)" if p_value < ALPHA else "fail to reject H0 (no evidence of a difference)")
    print("Size of the difference:", "negligible" if abs(cohens_d) < 0.2 else "small" if abs(cohens_d) < 0.5 else "medium or large",
          "(|d| < 0.2 is negligible even when p is tiny, because large samples make tiny gaps 'significant')")
else:
    print("No date column, so this test is skipped.")
````

##### Which statistical test have you done to obtain P-Value?

Welch's independent two-sample t-test (on log(1 + amount)), with Cohen's d as the effect size.

##### Why did you choose the specific statistical test?

There are two independent groups (weekday and weekend) and one numeric variable. Raw amounts are heavily skewed, so I compared the log-transformed amounts, which are close to normal (Chart 2). Welch's version does not assume the two groups have equal variance. With very large samples almost any gap gives a tiny p-value, so Cohen's d is reported to show whether the gap is big enough to matter.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**H0 (null):** Night-time (22:00-06:00) and daytime transaction amounts come from the same distribution.

**H1 (alternate):** Night-time and daytime transaction amounts come from different distributions.

#### 2. Perform an appropriate statistical test.

````python
# H2: are night-time transactions (22:00-06:00) different in size from daytime ones?
# Mann-Whitney U compares the two distributions without assuming a normal shape.
if "is_night" in viz:
    night = viz.loc[viz["is_night"] == 1, COL_AMOUNT]
    day = viz.loc[viz["is_night"] == 0, COL_AMOUNT]
    if len(night) > 20 and len(day) > 20:
        u_stat, p_value = stats.mannwhitneyu(night, day, alternative="two-sided")
        effect = u_stat / (len(night) * len(day))                              # chance a random night amount beats a random day amount
        print(f"Night: n={len(night):,}, median={night.median():,.2f}")
        print(f"Day:   n={len(day):,}, median={day.median():,.2f}")
        print(f"U = {u_stat:,.0f}, p-value = {p_value:.4g}, P(night > day) = {effect:.3f} (0.5 means no difference)")
        print("Conclusion:", "reject H0 (the distributions differ)" if p_value < ALPHA else "fail to reject H0 (no evidence of a difference)")
    else:
        print("Too few night or day transactions for a reliable test.")
else:
    print("No date column, so this test is skipped.")
````

##### Which statistical test have you done to obtain P-Value?

Mann-Whitney U test (two-sided).

##### Why did you choose the specific statistical test?

There are two independent groups and one numeric variable, but raw amounts are strongly right-skewed (not normal), so a t-test is a poor fit. Mann-Whitney compares the groups by ranks and makes no normality assumption. This hypothesis matters for the project, since an unusual-hour feature only helps if night-time spending really behaves differently.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**H0 (null):** The distribution of transaction amounts is the same across all spending categories.

**H1 (alternate):** At least one category has a different distribution of transaction amounts.

#### 2. Perform an appropriate statistical test.

````python
# H3: does the typical transaction amount differ across spending categories?
# Kruskal-Wallis is the multi-group version of Mann-Whitney (a rank-based alternative to one-way ANOVA).
groups = [g["log_amt"].values for _, g in viz[viz[COL_CATEGORY].isin(top_cats)].groupby(COL_CATEGORY) if len(g) > 20]
if len(groups) >= 2:
    h_stat, p_value = stats.kruskal(*groups)
    n_total = sum(len(g) for g in groups)
    epsilon_sq = (h_stat - len(groups) + 1) / (n_total - len(groups))          # effect size, between 0 and 1
    print(f"Groups compared: {len(groups)} categories, {n_total:,} transactions")
    print(f"H = {h_stat:.2f}, p-value = {p_value:.4g}, epsilon-squared = {epsilon_sq:.3f}")
    print("Conclusion:", "reject H0 (at least one category differs)" if p_value < ALPHA else "fail to reject H0 (no evidence of a difference)")
    print("Median log-amount per category:")
    print(viz[viz[COL_CATEGORY].isin(top_cats)].groupby(COL_CATEGORY)["log_amt"].median().sort_values().round(3).to_string())
else:
    print("Fewer than two categories with enough rows, so this test is skipped.")
````

##### Which statistical test have you done to obtain P-Value?

Kruskal-Wallis H test, with epsilon-squared as the effect size.

##### Why did you choose the specific statistical test?

There are more than two independent groups (categories) and one numeric variable. One-way ANOVA assumes roughly normal data with equal variances, which does not hold here, so the rank-based Kruskal-Wallis test is the safer choice. A significant result supports using category-relative amounts (z-score within category) as a feature.

### 2. Handling Outliers

````python
# Outliers are what this project is trying to find, so they are measured here but NOT removed.
check_cols = [c for c in ["log_amt", "distance_km", "gap_hours", "age"] if c in fe]
rows = []
for c in check_cols:
    q1, q3 = fe[c].quantile([0.25, 0.75])
    iqr = q3 - q1
    n_out = int(((fe[c] < q1 - 1.5 * iqr) | (fe[c] > q3 + 1.5 * iqr)).sum())   # the usual 1.5 x IQR rule
    rows.append({"feature": c, "outliers (1.5 x IQR)": n_out, "share of rows": f"{n_out / len(fe):.2%}"})
display(pd.DataFrame(rows))
print("No rows were removed or capped. Heavy skew is handled by the log transform instead (see Data Transformation).")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

No outlier removal or capping. Unusual transactions are exactly what the detector must find, so deleting them would delete the answer. Outliers are measured with the 1.5 x IQR rule, and heavy skew is reduced with a log transform, which keeps extreme values in the data but stops them from dominating the distances K-Means calculates.

### 3. Categorical Encoding

````python
# Low-cardinality columns -> one-hot. High-cardinality columns (merchant) -> frequency based "rarity" score.
ONEHOT_MAX = 20                                    # above this many distinct values one-hot would add too many columns
cat_cols = [COL_CATEGORY] + ([COL_GENDER] if COL_GENDER else [])
onehot_cols = [c for c in cat_cols if fe[c].nunique() <= ONEHOT_MAX]
freq_cols = [c for c in cat_cols if c not in onehot_cols] + ([COL_MERCHANT] if COL_MERCHANT else [])

encoded = (pd.get_dummies(fe[onehot_cols].astype(str), prefix=onehot_cols, dtype=int)
           if onehot_cols else pd.DataFrame(index=fe.index))
rarity_cols = []
for c in freq_cols:
    share = fe[c].map(fe[c].value_counts(normalize=True))
    fe[f"{c}_rarity"] = -np.log10(share)           # higher value = rarer value, so unusual merchants stand out
    rarity_cols.append(f"{c}_rarity")

print("One-hot encoded:", onehot_cols, f"-> {encoded.shape[1]} columns")
print("Frequency (rarity) encoded:", freq_cols)
````

#### What all categorical encoding techniques have you used & why did you use those techniques?

**One-hot encoding** for low-cardinality columns (category, gender): they have no natural order and few values, so one-hot adds a small number of columns. **Frequency (rarity) encoding** for merchant, which has too many values for one-hot: each merchant is replaced by how rare it is (-log10 of its share), so unusual merchants get higher values.

### 4. Feature Manipulation & Selection

#### 1. Feature Manipulation

````python
# New features that describe spending behaviour (each one is explained in the table in the next cell)
g = fe.groupby(COL_CATEGORY)["log_amt"]
fe["amt_cat_z"] = ((fe["log_amt"] - g.transform("mean")) / g.transform("std").replace(0, np.nan)).fillna(0)

if "hour" in fe:                                   # hour is circular: 23:00 and 00:00 are neighbours
    fe["hour_sin"] = np.sin(2 * np.pi * fe["hour"] / 24)
    fe["hour_cos"] = np.cos(2 * np.pi * fe["hour"] / 24)
if "gap_hours" in fe:
    fe["log_gap"] = np.log1p(fe["gap_hours"])
if "distance_km" in fe:
    fe["log_distance"] = np.log1p(fe["distance_km"])

merchant_rarity = [c for c in rarity_cols if c.startswith(str(COL_MERCHANT))]
candidates = [c for c in ["amt_cat_z", "log_amt", *merchant_rarity, "hour_sin", "hour_cos", "is_weekend", "is_night",
                          "log_gap", "is_first_txn", "log_distance", "age"] if c in fe]

# Look for pairs that carry almost the same information (|correlation| > 0.9)
corr = fe[candidates].corr().abs()
pairs = [(a, b, round(corr.loc[a, b], 2)) for i, a in enumerate(candidates) for b in candidates[i + 1:]
         if corr.loc[a, b] > 0.9]
print("Candidate numeric features:", candidates)
print("Highly correlated pairs (|r| > 0.9):", pairs if pairs else "none")
````

#### 2. Feature Selection

````python
# Unsupervised selection: no label is used. Remove near-constant features and one feature of each highly correlated pair.
# "Near-constant" = one single value covers more than 99.5% of the rows (independent of the feature's scale)
low_variance = [c for c in candidates if fe[c].round(6).value_counts(normalize=True).iloc[0] > 0.995]
redundant = {b for _, b, _ in pairs}                # of each pair, the one listed later is dropped
selected = [c for c in candidates if c not in low_variance and c not in redundant]

X = pd.concat([fe[selected], encoded], axis=1)      # final feature matrix (before scaling)
print("Dropped as near-constant:", low_variance or "none")
print("Dropped as redundant:", sorted(redundant) or "none")
print(f"Final feature matrix: {X.shape[0]:,} rows x {X.shape[1]} columns")

why = {
    "amt_cat_z": "amount compared with its own category: a 500 purchase is normal for travel, odd for coffee",
    "log_amt": "absolute size of the purchase on a log scale",
    "merchant_rarity": "rarely used merchants are less typical",
    "hour_sin": "time of day (circular, part 1)", "hour_cos": "time of day (circular, part 2)",
    "is_weekend": "weekend behaviour can differ", "is_night": "night-time purchases are less common",
    "log_gap": "burst of purchases or long silence since the card's previous transaction",
    "is_first_txn": "marks rows with no previous transaction", "log_distance": "far-from-home purchases",
    "age": "customer age, a behaviour context",
}
display(pd.DataFrame({"feature": selected,
                      "why it may help find anomalies": [why.get(f, "rarity of this value") for f in selected]}))
print(f"Plus {encoded.shape[1]} one-hot columns for: {onehot_cols}")
````

##### What all feature selection methods have you used  and why?

Unsupervised selection, because there is no target to select against: (1) a **near-constant filter** removes features where one value covers over 99.5% of rows, since they cannot separate transactions, and (2) a **correlation filter** (|r| > 0.9) drops one feature of each near-duplicate pair, since K-Means would otherwise count that information twice.

## ***7. ML Model Implementation***

````python
# Shared helpers for every model in this section: scoring, synthetic anomalies, and one evaluation function.
from sklearn.metrics import roc_auc_score, average_precision_score, adjusted_rand_score
from sklearn.neighbors import LocalOutlierFactor
from sklearn.model_selection import ParameterGrid

results = {}                                                   # model name -> metrics, filled as models are evaluated
has_labels = bool(COL_LABEL) and fe.loc[test_idx, COL_LABEL].nunique() > 1
y_test_arr = fe.loc[test_idx, COL_LABEL].values if has_labels else None

# Synthetic anomalies: copies of 1% of the test rows with the amount pushed 4-8 standard deviations upward.
# They tell us whether a model can find clear anomalies even when the data has no labels.
INJECT_COL = "amt_cat_z" if "amt_cat_z" in X_scaled.columns else X_scaled.columns[0]
n_inj = max(20, int(0.01 * len(X_test)))
X_inj = X_test.sample(n_inj, random_state=42).copy()
X_inj[INJECT_COL] += np.random.default_rng(42).uniform(4, 8, n_inj)
print(f"Test rows: {len(X_test):,} | injected anomalies: {n_inj} (feature shifted: {INJECT_COL}) | real labels available: {has_labels}")


# --- K-Means scoring: distance to own centroid, normalised inside the cluster, plus a small-cluster flag ---
def kmeans_ref(km, X_ref):
    """Learn from reference (training) data what a normal distance looks like in each cluster."""
    lab = km.predict(X_ref)
    d = km.transform(X_ref).min(axis=1)
    stats_by_cluster = pd.DataFrame({"c": lab, "d": d}).groupby("c")["d"].agg(["mean", "std"])
    share = pd.Series(lab).value_counts(normalize=True)
    return stats_by_cluster, share

def kmeans_score(km, X, ref, small_cut=0.01, weight=2):
    """Higher score = more unusual. Works on unseen rows because it reuses the training statistics."""
    stats_by_cluster, share = ref
    lab = km.predict(X)
    d = km.transform(X).min(axis=1)
    mu = stats_by_cluster["mean"].reindex(lab).to_numpy()
    sd = stats_by_cluster["std"].reindex(lab).replace(0, np.nan).fillna(1).to_numpy()
    small = (share.reindex(lab).fillna(0).to_numpy() < small_cut).astype(int)
    return (d - mu) / sd + weight * small


# --- Evaluation shared by all models ---
def evaluate(name, s_test, s_inj):
    """s_test / s_inj: anomaly scores of the test rows and of the injected rows (higher = more unusual)."""
    s_test, s_inj = np.asarray(s_test), np.asarray(s_inj)
    combined = np.concatenate([s_test, s_inj])
    cut = np.quantile(combined, 0.95)                                # flag the top 5% of everything
    pct_rank = combined.argsort().argsort() / (len(combined) - 1)
    out = {"Injected recall (top 5%)": float((s_inj >= cut).mean()),
           "Injected mean percentile": float(pct_rank[len(s_test):].mean())}
    if has_labels:
        order = np.argsort(-s_test)
        k1, k5 = max(1, int(0.01 * len(s_test))), max(1, int(0.05 * len(s_test)))
        out["ROC-AUC"] = roc_auc_score(y_test_arr, s_test)
        out["PR-AUC"] = average_precision_score(y_test_arr, s_test)
        out["Precision (top 1%)"] = float(y_test_arr[order[:k1]].mean())
        out["Recall (top 5%)"] = float(y_test_arr[order[:k5]].sum() / max(1, y_test_arr.sum()))
    results[name] = out
    return out

def plot_metrics(name):
    m = pd.Series(results[name])
    fig, ax = plt.subplots(figsize=(9, 4))
    bars = ax.barh(m.index[::-1], m.values[::-1], color="steelblue")
    ax.set_xlim(0, 1.05); ax.set_xlabel("Score (higher is better)")
    ax.set_title(f"Evaluation metric scores: {name}")
    for b, v in zip(bars, m.values[::-1]):
        ax.text(v + 0.01, b.get_y() + b.get_height() / 2, f"{v:.3f}", va="center")
    plt.tight_layout(); plt.show()
    if has_labels:
        print(f"Reference: a model with no skill gets PR-AUC of about {y_test_arr.mean():.3f} (the share of labelled anomalies).")
````

### ML Model - 1

````python
# ML Model 1: K-Means with an untuned starting value of k
BASE_K = 8
km_base = KMeans(n_clusters=BASE_K, n_init=10, random_state=42).fit(X_train)       # Fit the algorithm
ref_base = kmeans_ref(km_base, X_train)                                            # what "normal" distance is per cluster

s_test = kmeans_score(km_base, X_test, ref_base)                                   # Predict: anomaly score of unseen rows
s_inj = kmeans_score(km_base, X_inj, ref_base)

model1_name = f"K-Means (k={BASE_K}, default)"
sil = silhouette_score(X_train, km_base.labels_, sample_size=min(10_000, len(X_train)), random_state=42)
print(f"Silhouette score on the training data: {sil:.3f}")
print("Cluster sizes (share of training rows):")
print(pd.Series(km_base.labels_).value_counts(normalize=True).sort_index().round(3).to_string())
display(pd.Series(evaluate(model1_name, s_test, s_inj)).round(3).to_frame("score"))
````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

**Model: K-Means.** It groups transactions into k clusters around centroids. A normal transaction sits close to its own centroid. The anomaly score is the distance to that centroid, normalised inside its cluster, plus extra weight for members of tiny clusters. Metrics are in the chart below; labelled-data metrics (ROC-AUC, PR-AUC, precision, recall) appear only if the dataset has a label, and the injected-anomaly metrics always appear.

````python
# Visualizing evaluation Metric Score chart
plot_metrics(model1_name)
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ML Model 1 with hyperparameter tuning: choose k by elbow and silhouette, then check stability across random seeds.
# There is no target to optimise, so instead of GridSearchCV this is a manual grid over k scored by the silhouette.
tune_sample = X_train.sample(min(len(X_train), 50_000), random_state=42)           # a sample keeps the search fast
rows = []
for k in range(2, 13):
    km_k = KMeans(n_clusters=k, n_init=5, random_state=42).fit(tune_sample)
    sil_k = silhouette_score(tune_sample, km_k.labels_, sample_size=min(10_000, len(tune_sample)), random_state=42)
    rows.append({"k": k, "inertia": km_k.inertia_, "silhouette": sil_k})
tune = pd.DataFrame(rows)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(tune["k"], tune["inertia"], marker="o", color="steelblue")
axes[0].set_title("Elbow curve: inertia vs k"); axes[0].set_xlabel("k"); axes[0].set_ylabel("Inertia")
axes[1].plot(tune["k"], tune["silhouette"], marker="o", color="seagreen")
axes[1].set_title("Silhouette score vs k"); axes[1].set_xlabel("k"); axes[1].set_ylabel("Silhouette")
plt.tight_layout(); plt.show()

K_CHOSEN = None            # type a number here to override the automatic choice after reading both curves
allowed = tune[tune["k"] >= 3]                    # k = 2 usually wins on silhouette but gives clusters too coarse to profile
best_k = int(allowed.loc[allowed["silhouette"].idxmax(), "k"]) if K_CHOSEN is None else int(K_CHOSEN)
print(f"Chosen k = {best_k}")

km_best = KMeans(n_clusters=best_k, n_init=10, random_state=42).fit(X_train)
ref_best = kmeans_ref(km_best, X_train)
model1_tuned = f"K-Means (k={best_k}, tuned)"
evaluate(model1_tuned, kmeans_score(km_best, X_test, ref_best), kmeans_score(km_best, X_inj, ref_best))

# Stability check (the unsupervised counterpart of cross-validation): refit with other seeds and compare the groupings.
# Adjusted Rand index: 1 = identical clusters, 0 = no agreement beyond chance.
reference_labels = km_best.predict(X_test)
aris = [adjusted_rand_score(reference_labels,
                            KMeans(n_clusters=best_k, n_init=3, random_state=s).fit(tune_sample).predict(X_test))
        for s in range(1, 5)]
print(f"Stability across 4 other seeds (mean adjusted Rand index): {np.mean(aris):.3f}")

display(pd.DataFrame(results).T.loc[[model1_name, model1_tuned]].round(3))
plot_metrics(model1_tuned)
````

##### Which hyperparameter optimization technique have you used and why?

**Manual grid search over k (2 to 12)**, judged by the elbow curve and the silhouette score, because K-Means has no target to optimise and GridSearchCV needs one. For cross-validation, the grouping is refitted with other random seeds and compared with the adjusted Rand index (stability check).

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# Isolation Forest with a grid search. With no labels to optimise, the settings are scored by how highly the
# synthetic anomalies rank (mean percentile); real labels, if any, are only reported and never used for the choice.
fit_sample = X_train.sample(min(len(X_train), 50_000), random_state=42)
grid = ParameterGrid({"n_estimators": [100, 300], "max_samples": [256, "auto"], "max_features": [0.5, 1.0]})
rows = []
for params in grid:
    m = IsolationForest(random_state=42, n_jobs=-1, **params).fit(fit_sample)
    s_t, s_i = -m.score_samples(X_test), -m.score_samples(X_inj)
    pct = (np.concatenate([s_t, s_i]).argsort().argsort() / (len(s_t) + len(s_i) - 1))[len(s_t):].mean()
    rows.append({**params, "injected mean percentile": round(float(pct), 4)})
grid_results = pd.DataFrame(rows).sort_values("injected mean percentile", ascending=False)
display(grid_results)

best = grid_results.iloc[0].drop("injected mean percentile").to_dict()
best["n_estimators"] = int(best["n_estimators"]); best["max_features"] = float(best["max_features"])
if best["max_samples"] != "auto":
    best["max_samples"] = int(float(best["max_samples"]))
iso_best = IsolationForest(random_state=42, n_jobs=-1, **best).fit(X_train)
model2_tuned = "Isolation Forest (tuned)"
evaluate(model2_tuned, -iso_best.score_samples(X_test), -iso_best.score_samples(X_inj))
print("Best settings:", best)
display(pd.DataFrame(results).T.loc[[model2_name, model2_tuned]].round(3))
plot_metrics(model2_tuned)
````

##### Which hyperparameter optimization technique have you used and why?

**Grid search (ParameterGrid)** over n_estimators, max_samples and max_features. With no labels to optimise, each setting is scored by how highly the injected synthetic anomalies rank. Real labels, if present, are reported but not used to choose the settings, which avoids tuning on the test data.

#### 3. Explain each evaluation metric's indication towards business and the business impact pf the ML model used.

**Injected recall (top 5%)** and **mean percentile**: do obviously odd transactions end up at the top of the review list? **Precision (top 1%)**: of the transactions an auditor reviews first, how many are truly unusual? A low value means wasted review time. **Recall (top 5%)**: how many of the truly unusual transactions are caught within a review budget of 5% of transactions? A low value means missed problems. **PR-AUC**: precision and recall over every threshold, the right summary for rare anomalies. **ROC-AUC**: how well the score separates normal from anomalous overall. Accuracy is deliberately not used, because with under 1% anomalies a model that flags nothing would score over 99%.

### ML Model - 3

````python
# ML Model 3: Local Outlier Factor. It compares each point's local density with its neighbours' density.
# It is slow on big data, so it is fitted on a sample of the training rows.
lof_sample = X_train.sample(min(len(X_train), 20_000), random_state=42)
lof_base = LocalOutlierFactor(n_neighbors=20, novelty=True, n_jobs=-1).fit(lof_sample)     # novelty=True allows scoring unseen rows
model3_name = "Local Outlier Factor (k=20)"
s_test = -lof_base.score_samples(X_test)
s_inj = -lof_base.score_samples(X_inj)
display(pd.Series(evaluate(model3_name, s_test, s_inj)).round(3).to_frame("score"))
````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

**Model: Local Outlier Factor.** It compares the density around a transaction with the density around its nearest neighbours. A point in a sparse area surrounded by dense areas gets a high score. It catches local anomalies that global methods can miss, but it is slow on large data, so it is fitted on a sample.

````python
# Visualizing evaluation Metric Score chart
plot_metrics(model3_name)
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# Local Outlier Factor: grid search over the neighbourhood size, scored by the injected anomalies as before.
rows = []
for nn in [10, 20, 50, 100]:
    m = LocalOutlierFactor(n_neighbors=nn, novelty=True, n_jobs=-1).fit(lof_sample)
    s_t, s_i = -m.score_samples(X_test), -m.score_samples(X_inj)
    pct = (np.concatenate([s_t, s_i]).argsort().argsort() / (len(s_t) + len(s_i) - 1))[len(s_t):].mean()
    rows.append({"n_neighbors": nn, "injected mean percentile": round(float(pct), 4)})
lof_grid = pd.DataFrame(rows).sort_values("injected mean percentile", ascending=False)
display(lof_grid)

best_nn = int(lof_grid.iloc[0]["n_neighbors"])
lof_best = LocalOutlierFactor(n_neighbors=best_nn, novelty=True, n_jobs=-1).fit(lof_sample)
model3_tuned = f"Local Outlier Factor (k={best_nn}, tuned)"
evaluate(model3_tuned, -lof_best.score_samples(X_test), -lof_best.score_samples(X_inj))
display(pd.DataFrame(results).T.loc[[model3_name, model3_tuned]].round(3))
plot_metrics(model3_tuned)
````

##### Which hyperparameter optimization technique have you used and why?

**Grid search over n_neighbors**, the one setting that matters most for LOF, scored by how highly the injected synthetic anomalies rank (needs no real labels).
